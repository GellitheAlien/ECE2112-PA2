## A. Reproducible Normalization Problem

```python
import numpy as np
```

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
X
```

```python
Xsum = np.sum(X)
Xsum
```

```python
mean = Xsum/ X.size
mean
```

```python
std = np.std(X[0:6,0:6])
std
```

```python
X_normalized = (X - mean)/std
X_normalized
```

```python
np.save('X_normalized', X_normalized)
```

## B. Cubes Divisible by 4 Problem

```python
C = np.power(np.arange(1, 101, 1), 3).reshape(10,10)
C
```

```python
div_by_4 = C[C % 4 == 0]
div_by_4
```

```python
np.save('div_by_4', div_by_4)
```

## C. Above-Mean Square Problem

```python
S = np.power(np.arange(1, 37, 1), 2).reshape(6,6)
S
```

```python
S_mean = np.mean(S)
S_mean
```

```python
above_mean = S[S > S_mean]
above_mean
```

```python
np.save('above_mean', above_mean)
```
